## 13. Standard Data Types & Mutability

| Type | Category | Mutability | Example |
|---|---|---|---|
| `int` | Numeric | Immutable | `x = 42` |
| `float` | Numeric | Immutable | `pi = 3.14` |
| `complex` | Numeric | Immutable | `z = 3+4j` |
| `str` | Text | Immutable | `s = "hello"` |
| `bool` | Boolean | Immutable | `flag = True` |
| `tuple` | Sequence | Immutable | `t = (1,2,3)` |
| `list` | Sequence | **Mutable** | `l = [1,2,3]` |
| `dict` | Mapping | **Mutable** | `d = {"a":1}` |
| `set` | Set | **Mutable** | `s = {1,2,3}` |
| `frozenset` | Set | Immutable | `fs = frozenset({1,2})` |
| `NoneType` | Null | Immutable | `x = None` |

> **Immutable** = once created, value cannot change in-place (creates a new object).  
> **Mutable** = you can modify the object in-place.  
> *Analogy:* Immutable = printed book (reprint to change). Mutable = whiteboard (erase & rewrite).

## **sys.getsizeof() tells you how many bytes of memory the Python object occupies, including Python's internal bookkeeping.bold text**

```python
import sys

# ----------------Integer (int) -------------
a= 42
print(type(a))
print(isinstance(a, int))
print(sys.getsizeof(a))

# ----------------Float (float) -------------
b = 3.14
print(type(b))
print(isinstance(b, float))
print(sys.getsizeof(b))

# ----------------Complex (complex) -------------
c = 3 + 4j
print(type(c))
print(isinstance(c, complex))
print(sys.getsizeof(c))
```

# **Structural pattern matching**

It was introducted in python 3.10. Like a switch statement: match a value against pattern.

```python
def describe_type(obj):
  match obj:
    case int():
      return 'It is an integer'
    case float():
      return 'It is a float'
    case str():
      return 'It is a string'
    case list():
      return 'It is a list'
    case _:
      return 'Unknown type'

print(describe_type(42.00))
print(describe_type(3.14))
print(describe_type('Hello'))
print(describe_type([1,2,3]))
print(describe_type(True))
```

```python
def day_type(day):
  match day:
    case 'Saturday' | 'Sunday':
      return 'Weekend'
    case 'Monday' | 'Tuesday' | 'Wednesday' | 'Thursday' | 'Friday':
      return 'Weekday'
    case _:
      return 'Invalid day'

print(day_type('Saturday'))
print(day_type('Monday'))
print(day_type('Invalid'))
```
