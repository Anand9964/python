#Python Optimization

Reusing frequently used immutable objects:

Saves memory
Improves performance
Avoids creating duplicate objects unnecessarily

Since integers are immutable, sharing them is completely safe.

$Modulo

```python
name = "Pravin"
score = 98.567
subjects = int(input("Enter the number of subjects: "))

print("Student: %s, Score: %f, Subject: %a" %(name, score, subjects))

print("Student: %s, Score: %.2f, Subject: %a" %(name, score, subjects))

print("Student: %s, Score: %8.2f, Subject: %a" %(name, score, subjects))

print("Student: %s, Score: %05d, Subject: %a" %(name, score, subjects))

```

```python
# Method 2: str.format()
name = "Anand"
score = 98.567

print("Student: {}, Score: {}".format(name, score))

print("value exchanging index wise, Student: {1}, Score: {0}".format(name, score))

print("Student: {}, Score: {:.2f}".format(name, score))
```

```python
# Method 3: f-string
name = "Anand"
score = 98.567
items = ["Python","AI","ML"]

print(f"Student: {name}, Score: {score}")
print(f"Student: {name}, Score: {score:.2f}")
print(f"Student: {name}, Score: {score:.2f}, Subjects: {items}")

print(f"Upper Case: {name.upper()}")
print(f"Lower Case: {name.lower()}")

```

## 13. Standard Data Types & Mutability

| Type | Category | Mutability | Example |
|---|---|---|---|
| `int` | Numeric | Immutable | `x = 42` |
| `float` | Numeric | Immutable | `pi = 3.14` |
| `complex` | Numeric | Immutable | `z = 3+4j` |
| `str` | Text | Immutable | `s = "hello"` |
| `bool` | Boolean | Immutable | `flag = True` |
| `tuple` | Sequence | Immutable | `t = (1,2,3)` |
| `list` | Sequence | **Mutable** | `l = [1,2,3]` |
| `dict` | Mapping | **Mutable** | `d = {"a":1}` |
| `set` | Set | **Mutable** | `s = {1,2,3}` |
| `frozenset` | Set | Immutable | `fs = frozenset({1,2})` |
| `NoneType` | Null | Immutable | `x = None` |

> **Immutable** = once created, value cannot change in-place (creates a new object).  
> **Mutable** = you can modify the object in-place.  
> *Analogy:* Immutable = printed book (reprint to change). Mutable = whiteboard (erase & rewrite).

```python
import sys

# ----------------Integer (int) -------------
a= 42
print(type(a))
print(isinstance(a, int))
print(sys.getsizeof(a))

# ----------------Float (float) -------------
b = 3.14
print(type(b))
print(isinstance(b, float))
print(sys.getsizeof(b))

# ----------------Complex (complex) -------------
c = 3 + 4j
print(type(c))
print(isinstance(c, complex))
print(sys.getsizeof(c))
```
