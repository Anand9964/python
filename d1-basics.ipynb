## 13. Standard Data Types & Mutability

| Type | Category | Mutability | Example |
|---|---|---|---|
| `int` | Numeric | Immutable | `x = 42` |
| `float` | Numeric | Immutable | `pi = 3.14` |
| `complex` | Numeric | Immutable | `z = 3+4j` |
| `str` | Text | Immutable | `s = "hello"` |
| `bool` | Boolean | Immutable | `flag = True` |
| `tuple` | Sequence | Immutable | `t = (1,2,3)` |
| `list` | Sequence | **Mutable** | `l = [1,2,3]` |
| `dict` | Mapping | **Mutable** | `d = {"a":1}` |
| `set` | Set | **Mutable** | `s = {1,2,3}` |
| `frozenset` | Set | Immutable | `fs = frozenset({1,2})` |
| `NoneType` | Null | Immutable | `x = None` |

> **Immutable** = once created, value cannot change in-place (creates a new object).  
> **Mutable** = you can modify the object in-place.  
> *Analogy:* Immutable = printed book (reprint to change). Mutable = whiteboard (erase & rewrite).

```python
a = int(input("enter integer value"))
b = float(input("enter float value"))

print(id(a))
print(id(b))

a = 14
b = 16.23

print(id(a))
print(id(b))
```

```python
import sys

# ----------------Integer (int) -------------
a= 42
print(type(a))
print(isinstance(a, int))
print(sys.getsizeof(a))

# ----------------Float (float) -------------
b = 3.14
print(type(b))
print(isinstance(b, float))
print(sys.getsizeof(b))

# ----------------Complex (complex) -------------
c = 3 + 4j
print(type(c))
print(isinstance(c, complex))
print(sys.getsizeof(c))
```

## 15. Collections — Strings, Tuples, Lists, Dicts, Sets

| Type | Ordered? | Unique? | Mutable? | Indexed? | Best For |
|---|---|---|---|---|---|
| `str` | ✅ | ❌ | ❌ | ✅ | Text, messages |
| `tuple` | ✅ | ❌ | ❌ | ✅ | Fixed records, GPS coords |
| `list` | ✅ | ❌ | ✅ | ✅ | Dynamic collections, stacks |
| `dict` | ✅(3.7+) | Keys only | ✅ | By key | Lookup tables, JSON, config |
| `set` | ❌ | ✅ | ✅ | ❌ | Deduplication, membership tests |

```python
# List
fruits = ["apple","banana","mango"]
print(fruits)
print(type(fruits))
print(len(fruits))
print(fruits[0])
print(fruits[1])
print(fruits[2])
print(fruits[-1])
print(fruits[-2])
print(fruits[-3])

fruits[0] = "orange1"
print(fruits)
fruits.append("orange2")
print(fruits)
fruits.remove("banana")
print(fruits)
fruits.insert(1,"banana")
print(fruits)
fruits.sort()
print(fruits)
fruits.reverse()
print(fruits)
```

```python
# Tuples : it's like list, but once created it cannot be changed
fruits = ("apple","banana","mango")
print(fruits)
print(type(fruits))
print(len(fruits))
print(fruits[0])
print(fruits[1])
print(fruits[2])

fruits[0] = "orange1"
print(fruits)
```

```python
# Dictionary map keys to its values

students = {
    "name": "Anand",
    "age": 36,
    "city": "Bengaluru"
}

print(students)
print(type(students))
print(len(students))
print(students["name"])
print(students["age"])
print(students["city"])

students["age"] = 37
print(students)


# Map of map
students = {
    "Students1":{
        "name": "Anand",
        "age": 36,
        "city": "Bengaluru"
    },
    "Students2":{
        "name": "Shwetha",
        "age": 33,
        "city": "Tumkur"
    }
}

print(students)
print(type(students))
print(len(students))
print(students["Students1"])
print(students["Students2"])
print(students["Students1"]["name"])
```

```python
#Sets
numbers = {1,2,3,4,5,5,3,6}
print(numbers)
print(type(numbers))
print(len(numbers))

numbers.add(7)
print(numbers)

numbers.remove(3)
print(numbers)

numbers.add(3)
print(numbers)

#
```

```python
# indexing and slicing
s = "Hello, Python World!!"
print(s)
print(len(s))
print(s[0]) #First char
print(s[-1]) #Second char

print(s[0:5])
print(s[::2]) # printing every second character
print(s[::-1]) # reversing sring

#String methods
print(s.lower())
print(s.upper())

print("--------------")

print(s.replace("Python","AI"))
print(s.split(","))
print(" Anand ".strip())
print(s.strip())
```
