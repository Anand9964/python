# Handling Missing Values with Mean, Median and Mode

```python
import pandas as pd
import numpy as np

# Sample dataset with missing values
data = {
    'Age':    [25, np.nan, 30, np.nan, 35, 40, np.nan],
    'Salary': [50000, 60000, np.nan, 55000, np.nan, 70000, 65000],
    'City':   ['NY', 'LA', np.nan, 'NY', 'LA', np.nan, 'NY']
}

df = pd.DataFrame(data)
print('Original DataFrame:')
print(df)
print('\nMissing values:\n', df.isnull().sum())
```

## Fill with Mean (numerical — symmetric data)

```python
df_mean = df.copy()
df_mean['Age']    = df_mean['Age'].fillna(df_mean['Age'].mean())
df_mean['Salary'] = df_mean['Salary'].fillna(df_mean['Salary'].mean())
print('After Mean Imputation:')
print(df_mean)
```

## Fill with Median (numerical — skewed data)

```python
df_median = df.copy()
df_median['Age']    = df_median['Age'].fillna(df_median['Age'].median())
df_median['Salary'] = df_median['Salary'].fillna(df_median['Salary'].median())
print('After Median Imputation:')
print(df_median)
```

## Fill with Mode (categorical or discrete data)

```python
df_mode = df.copy()
df_mode['City'] = df_mode['City'].fillna(df_mode['City'].mode()[0])
df_mode['Age']  = df_mode['Age'].fillna(df_mode['Age'].mode()[0])
print('After Mode Imputation:')
print(df_mode)
```

## Summary: When to use what?

| Strategy | Best for |
|----------|----------|
| **Mean** | Numerical, normally distributed, no outliers |
| **Median** | Numerical, skewed distribution or outliers present |
| **Mode** | Categorical or discrete columns |
